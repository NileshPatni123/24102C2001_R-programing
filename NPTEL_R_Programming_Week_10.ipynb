## SLOT 1 – String Substitution using `sub()`
The notes explain that `sub()` replaces only the **first occurrence** of a matching substring.

```r
y <- "Number of participants: 25"
sub("25", "30", y)
```

## SLOT 2 – `sub()` with repeated text
Only the first `Mr. Singh` is replaced.

```r
y <- "Mr. Singh is the smart one. Mr. Singh is funny, too."
sub("Mr. Singh", "Professor Jha", y)
```

## SLOT 3 – Global Substitution using `gsub()`
`gsub()` replaces **all occurrences** of the matching substring.

```r
y <- "Mr. Singh is the smart one. Mr. Singh is funny, too."
gsub("Mr. Singh", "Professor Jha", y)
```

## SLOT 4 – `grep()` with `value = TRUE`
Returns the actual matching strings.

```r
str <- c("R Course", "exercises", "include examples of R language")
grep("ex", str, value = TRUE)
```

## SLOT 5 – `grep()` with `value = FALSE`
Returns the **indices/positions** of matching elements.

```r
str <- c("R Course", "exercises", "include examples of R language")
grep("ex", str, value = FALSE)
```

## SLOT 6 – `grep()` and `ignore.case`
Case-sensitive matching is used when `ignore.case = FALSE`.

```r
str <- c("R Course", "exercises", "include examples of r language", "in R software.")
grep("R", str, ignore.case = FALSE, value = TRUE)
```

## SLOT 7 – Case-insensitive `grep()`
With `ignore.case = TRUE`, both upper- and lower-case `R` are matched.

```r
grep("R", str, ignore.case = TRUE, value = TRUE)
```

## SLOT 8 – `grep()` returning indices
Compare the indices for case-insensitive and case-sensitive matching.

```r
grep("R", str, ignore.case = TRUE, value = FALSE)
```

```r
grep("R", str, ignore.case = FALSE, value = FALSE)
```

## SLOT 9 – `grep()` on combined strings
The notes use `c(x, y)` to combine two strings and then search them.

```r
x <- "R course 24.07.2021"
y <- "Number of participants: 25"
c(x, y)
grep("our", c(x, y))
```

## SLOT 10 – `grep()` for `Num`

```r
x <- "R course 24.07.2022"
y <- "Number of participants: 50"
grep("Num", c(x, y))
```

## SLOT 11 – `grepl()`
`grepl()` gives a logical result (`TRUE`/`FALSE`) for each element.

```r
str <- c("R Course", "exercises", "include examples of R language")
grepl("R", str)
```

## SLOT 12 – `grepl()` for another pattern

```r
grepl("ex", str)
```

# PART C – COMBINING AND MERGING DATA FRAMES

The notes give three main techniques:
- `cbind()` → combine columns side-by-side
- `merge()` → join using a common column
- `rbind()` → stack rows vertically

## SLOT 31 – Create `df1`

```r
df1 <- data.frame(
  state = c("UP", "MP", "AP", "JK"),
  popnsize = c(1000, 2000, 3000, 4000)
)
df1
```

## SLOT 32 – Create `df2`

```r
df2 <- data.frame(
  state = c("UP", "MP", "AP", "JK"),
  samplesize = c(100, 200, 300, 400),
  surveycompleted = c("Yes", "No", "Yes", "No")
)
df2
```

## SLOT 33 – `cbind()`
Combines the two data frames side-by-side.

```r
cbind(df1, df2)
```

## SLOT 34 – `merge()` using common column `state`
The common column is used to join the two data frames.

```r
merge(df1, df2, by = "state")
```

## SLOT 35 – Create `df11` for vertical combining

```r
df11 <- data.frame(
  state = c("UP", "MP", "AP", "JK"),
  popnsize = c(1000, 2000, 3000, 4000)
)
df11
```

## SLOT 36 – Create `df22`

```r
df22 <- data.frame(
  state = c("Bihar", "Delhi", "Punjab"),
  popnsize = c(100, 200, 300)
)
df22
```

## SLOT 37 – `rbind()`
Stacks the rows of `df22` below `df11`.

```r
rbind(df11, df22)
```
