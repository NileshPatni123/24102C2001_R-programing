## Lecture 36 – `cat()` / Print and Formatting

```r
# print() prints one object at a time
print("The zero occurs at")
print(2*pi)
print("radians")

# cat() combines multiple items into continuous output
cat("The zero occurs at", 2*pi, "radians.", "\n")

# Date example
d = date()
cat("Today's date is:", d, "\n")

# sep example
x = 1:10
x
cat(x, sep = " ++ ")
cat("\n")
cat(x, sep = " / ")
cat("\n")

# Formatting numbers
x = 7
cat("The square of", x, "is", x^2, "!\n")
cat("The square root of", x, "is approximately",
    format(sqrt(x), digits=3), "\n")

# Vector example
evenno = c(2,4,6,8,10)
evenno
cat("The first few even numbers are:", evenno, "...\n")

# fill and labels
x = 1:10
cat(x, fill = 2, labels = paste("(", letters[1:10], "):"))
```

## Lecture 37 – `paste()` and `paste0()`

```r
# paste() concatenates strings
paste(1:12)
as.character(1:12)

paste("Everybody", "loves", "R Programming.")
paste("Everybody", "loves", "R Programming.", sep="*")
paste("Everybody", "loves", "R Programming.", sep="===")

# Vectorized paste
names = c("Prof. Singh", "Mr. Venkat", "Dr. Jha")
names
paste(names, "is", "a good", "person.")

# collapse
paste(names, "is", "a good", "person.", collapse=", and ")

# Operations with strings
paste(1, " is first", 2, " is second", 3, " is third", sep="#")
paste(1, " is first ", 2, " is second ", 3, " is third ", sep="#")

# Create a vector of strings
x = paste("Ex", 1:5, sep="_")
x
x[1]
x[2]
x[3]
x[5]

# collapse creates one string
x = paste("Ex", 1:5, sep="_", collapse="")
x
x[1]

# paste0
paste0(1:10)
paste(1:10)

paste0(1:10, c("st", "nd", "rd", rep("th", 7)))
paste(1:10, c("st", "nd", "rd", rep("th", 7)))
```

## Lecture 38 – `strsplit()`

```r
# Split a string using !
x = "The&!syntax&!of&!paste&!is!&available!&inthe online-help"
x
strsplit(x, split="!")

# Split using &!
x = "The&!syntax&!of&!paste&!is!&available!&inthe online-help"
x
strsplit(x, split="&!")

# Split using !&
x = "The&!syntax&!of&!paste&!is!&available!&inthe online-help"
y = strsplit(x, split = "!&")
y
y[[1]][1]
y[[1]][2]
y[[1]][3]

# Convert dates into a matrix
dates = c("2020-07-24", "2021-08-25", "2022-09-26", "2023-10-27")
datesplt = strsplit(dates, "-")
datesplt

# Character matrix
datemat = matrix(unlist(datesplt), nrow = 4, ncol=3, byrow=TRUE)
datemat

# Numeric matrix
datematrix = matrix(as.numeric(unlist(datesplt)),
                    nrow = 4, ncol=3, byrow=TRUE)
datematrix

# Split a word into individual characters
strsplit("Shalabh", split="")
```

## Lecture 39 – String Manipulation

```r
# nchar(): number of characters
x = "R course 24.07.2022"
y = "Number of participants: 25"

nchar(x)
nchar(y)

# nzchar(): whether strings are non-empty
nzchar(x)
nzchar(y)

# nzchar() with a vector
x = c("Apple", "Banana", "Cake")
x
nzchar(x)

y = c("Apple", "", "Cake")
y
nzchar(y)

# nchar() with vectors
x = c("Apple", "Banana", "Cake")
nchar(x)

y = c(2, 4, 6)
nchar(y)

z = c(11, 222, 3333)
nchar(z)

z1 = c(1.1, 2.22, 3.333)
nchar(z1)

# Case conversion
x = "R course will start from 24.07.2022"
toupper(x)

z = "INDIAN INSTITUTE OF TECHNOLOGY"
tolower(z)
```

## Quick Reference

- `cat()` – combines and prints multiple items.
- `paste()` – joins strings, with `sep` between inputs and optional `collapse` across results.
- `paste0()` – equivalent to `paste(..., sep = "")`.
- `strsplit()` – splits character vectors using a delimiter/pattern.
- `nchar()` – counts characters.
- `nzchar()` – checks whether strings are non-empty.
- `tolower()` / `toupper()` – changes alphabetic case.
